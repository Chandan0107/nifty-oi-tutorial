# NIFTy for Optical Interferometry — a tutorial

A step-by-step tutorial on Bayesian image reconstruction for optical/IR long-baseline
interferometry with [NIFTy8](https://ift.pages.mpcdf.de/nifty/).

| Notebook | Topic |
|---|---|
| `01_dataload` | Read an OIFITS file (VIS, VIS2, T3) into a dictionary |
| `02_interferometry_response` | `InterferometryResponse`: a NIFTy `LinearOperator` for image → visibilities using `ducc0.wgridder` |
| `03_sky_models` | Synthetic sky models: Gaussian, elongated Gaussian, uniform disk, ring |
| `04_nufft_check` | Sanity check: `finufft` NUFFT vs `ducc0` `dirty2vis` |
| `05_bayesian_reconstruction` | Priors, |V|², bispectrum (closure phase) likelihoods and MGVI inference with `ift.optimize_kl` |
